```python
import numpy as np
import matplotlib.pyplot as plt
import arviz as az

from src.data.data_handler import DataHandler
from src.strategies.basket_strategy import BasketStrategy
from src.models.hierarchical_bayesian import HierarchicalBayesianModel
```

```python
files = {
    "AAPL": "../../data/AAPL.csv",
    "TSLA": "../../data/TSLA.csv",
    "MSFT": "../../data/MSFT.csv"
}

dh = DataHandler(files)
returns = dh.get_returns()
```

```python
strategy = BasketStrategy(model_type="hierarchical")
result = strategy.run(returns)

posterior_means = result["posterior_means"]
weights = result["weights"]

posterior_means, weights
```

```python
plt.figure(figsize=(8,5))
plt.bar(posterior_means.keys(), posterior_means.values())
plt.title("Hierarchical Bayesian Expected Returns")
plt.ylabel("Expected Daily Return")
plt.show()
```
